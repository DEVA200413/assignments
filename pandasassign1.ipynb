```python
import numpy as np
import pandas as pd
```

```python
data=("https://raw.githubusercontent.com/laxmimerit/All-CSV-ML-Data-Files-Download/master/IMDB-Movie-Data.csv")
df=pd.read_csv(data)
```

```python
print(df.shape)
print(df.head())
print(df.columns)
print(df.dtypes)
print(df.isnull().sum())
```

```python
print(df.duplicated().sum())
df=df.drop_duplicates()
```

```python
df=df.rename(columns={'Title':'movie title'})
```

```python
print(df['Genre'].unique())
```

```python
df['Genre_count']=df['Genre'].str.split(',').str.len()
print(df[['Genre','Genre_count']].head())
```

```python
print(df['Revenue (Millions)'].isnull().sum())
print(df['Metascore'].isnull().sum())
```

```python
df['Revenue (Millions)']=df['Revenue (Millions)'].fillna(df['Revenue (Millions)'].mean())
df['Metascore']=df['Metascore'].fillna(df['Metascore'].mean())
```

```python
top5_movies = df.sort_values(by='Rating', ascending=False).head(5)
print(top5_movies[['movie title', 'Rating']])
```

```python
highest_revenue = df.sort_values(by='Revenue (Millions)', ascending=False).head(1)
print("Highest Revenue Movie:")
print(highest_revenue[['movie title', 'Revenue (Millions)']])

lowest_revenue = df.sort_values(by='Revenue (Millions)', ascending=True).head(1)
print("Lowest Revenue Movie:")
print(lowest_revenue[['movie title', 'Revenue (Millions)']])
```

```python
# Calculate the average rating for each director
avg_rating_by_director = df.groupby('Director')['Rating'].mean().reset_index()
print(avg_rating_by_director)
```

```python
movies_per_year = df['Year'].value_counts()
print(movies_per_year)
```

```python
filtered_movies = df[(df['Rating'] > 8) & (df['Revenue (Millions)'] > 100)]
print(filtered_movies[['movie title', 'Rating', 'Revenue (Millions)']])
```

```python
df.groupby('Director')[['Rating', 'Revenue (Millions)']].mean()
```

```python
df.groupby('Year')['Rating'].agg(['count', 'mean'])
```

```python
df['Profitability'] = df['Revenue (Millions)'] / df['Runtime (Minutes)']
top10_profitable = df.sort_values(by='Profitability', ascending=False).head(10)
print(top10_profitable[['movie title', 'Profitability']])
```

```python
import numpy as np

df['Rating_Category'] = np.select(
    [df['Rating'] < 6, (df['Rating'] >= 6) & (df['Rating'] < 7.5)],
    ['Poor', 'Average'],
    default='Excellent'
)

print(df[['movie title', 'Rating', 'Rating_Category']].head(10))
```

```python
def calculate_score(row):
    return row['Rating'] * 10 + row['Revenue (Millions)']

df['Score'] = df.apply(calculate_score, axis=1)

print(df[['movie title', 'Rating', 'Revenue (Millions)', 'Score']].head(10))
```
