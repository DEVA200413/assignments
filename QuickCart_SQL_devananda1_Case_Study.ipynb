# QuickCart – SQL Case Study

## 1. Connect to the SQLite database
Keep `QuickCartDB.db` in the same folder as this notebook.

```python
import sqlite3
import pandas as pd

db_path = "QuickCart_DB.db"
conn = sqlite3.connect(db_path)
conn.execute("PRAGMA foreign_keys = ON;")
print("Connected to QuickCart!")
```

## 2. Explore the database
List the available tables.

```python
tables = pd.read_sql_query(
    "SELECT name FROM sqlite_master WHERE type='table';", conn
)
display(tables)
```

### Inspect table structures

```python
for table in ["Customers", "Products", "Orders", "Order_Items", "Payments"]:
    print("\\n---", table, "---")
    display(pd.read_sql_query(f"PRAGMA table_info({table});", conn))
```

### Preview records

```python
for table in ["Customers", "Products", "Orders", "Order_Items", "Payments"]:
    print("\\n---", table, "---")
    display(pd.read_sql_query(f"SELECT * FROM {table} LIMIT 5;", conn))
```

```python
query = """
SELECT Customers.customer_name, SUM(Order_Items.quantity * Order_Items.selling_price) AS total_spent FROM Customers JOIN Orders ON Customers.customer_id = Orders.customer_id JOIN Order_Items
ON Orders.order_id = Order_Items.order_id
GROUP BY Customers.customer_name
ORDER BY total_spent DESC
LIMIT 1;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT Customers.* FROM Customers LEFT JOIN Orders ON Customers.customer_id = Orders.customer_id WHERE Orders.order_id IS NULL;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT Customers.customer_name, COUNT(Orders.order_id) AS total_orders FROM Customers JOIN Orders ON Customers.customer_id = Orders.customer_id GROUP BY Customers.customer_name
HAVING COUNT(Orders.order_id) >= 2;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT AVG(total_amount) AS average_order_value
FROM( SELECT order_id, SUM(quantity * selling_price) AS total_amount FROM Order_Items
    GROUP BY order_id
) AS order_values;
"""

pd.read_sql_query(query, conn)
```
